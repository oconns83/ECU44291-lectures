# pandas cheat sheet: reading and using data

**ECU44291 Quantitative Macroeconomics.**

pandas reads a data file into a **table** (a `DataFrame`) and lets you pick out columns, rows and summaries **by name**. NumPy is for the model; pandas is for the data. Every example uses the course data and lines from the calibration notebook.

## At a glance

| You see | It means | Example from our code |
|---|---|---|
| `pd.read_csv(path)` | read a CSV file (or a URL) into a DataFrame | `pd.read_csv(DATA + "gdp_gni_star.csv")` |
| `df.col` or `df["col"]` | one column: a **Series** (an array with row labels) | `cso.k_over_gdp` |
| `df[["a", "b"]]` | several columns: still a DataFrame | `pension[["value", "period"]]` |
| `.iloc[i]` | the entry at **position** `i`, counting from 0 | `share.alpha.iloc[-1]`: the last |
| `.loc[label]` | the entry with **label** `label` | `persons.loc[21:80]`: ages 21 to 80 |
| `df[condition]` | only the rows where the condition is `True` | `share[share.year <= 2014]` |
| `&`, \|, `~` | and, or, not, inside a condition; brackets round each part | `(d.scenario == "M1") & (d.ratio == "old")` |
| `.set_index("col")` | use a column as the row labels | `pop.set_index("age")` |
| `.sum()`, `.mean()`, `.min()`, `.max()` | a summary of a column | `early.alpha.mean()` |
| `df["new"] = ...` | add a column, computed from others | `cso["dwellings_share"] = ...` |
| `.merge(other, on="year")` | line two tables up by a column | `cso.merge(share, on="year")` |
| `.to_numpy()` | the bare NumPy array, for the model | `persons.loc[21:80].to_numpy()` |
| `.head()`, `.tail()`, `.shape`, `.columns` | first rows, last rows, (rows, columns), column names | `gdp.shape` |

## Different in pandas

| | In pandas | The trap |
|--|-----|-----|
| Plain brackets | `s[5]` on a column looks for the row **labelled** 5, not the sixth row. | `s[-1]` is a `KeyError`: no row is labelled $-1$. The last row is `.iloc[-1]`. |
| Label slices | **`.loc[21:65]` includes 65.** It goes by label, from 21 to 65. | Every other slice in Python stops before its end, including `.iloc[21:65]`. |
| Conditions | `&` and \|, with brackets: `(pop.age >= 21) & (pop.age <= 65)`. | `and`/`or` on a column is an error: *the truth value of a Series is ambiguous*. |
| Arithmetic | Two Series are lined up **by label**, not by position. | Labels that do not match give `NaN`, with no error. |
| Filtered rows | A filtered table keeps the old row labels: row 0 may be gone. | `.loc[0]` can fail where `.iloc[0]` works. |
| Changing a part | Change a filtered table only after `.copy()`. | Otherwise a *SettingWithCopyWarning*, and the change may not stick. |
| Dot or brackets | `df.col` and `df["col"]` are the same column. | Creating a column needs brackets: `df["new"] = ...`. |
